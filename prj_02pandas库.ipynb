### 创建Series

```python
import pandas as pd
# 使用列表创建
s = pd.Series([10, 2, 3,  4, 5])
print(s)

# 自定义索引与name
s = pd.Series([10, 2, 3, 4, 5], index = [6, 'a', 8, 9, 100], name = '名字')
print(s)
```

```python
# 使用字典来创建
s = pd.Series({'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5}, name = 'name111')
print(s)

# 从原序列按照index获取对应组合创建新series
s1 = pd.Series(s, index = ['a', 'c'])
print(s1)
```

### 属性

```python
# 索引与值
print(s.index, s.values)

# 形状, 维数, 元素个数, 元素类型
print(s.shape, s.ndim, s.size, s.dtype)

# 名字
print(s.name)

# 显式索引(标签)及切片, 隐式索引(下标), 单个元素访问(不可切片)
print(s.loc['a'], '\n', s.loc['a': 'c'], s.iloc[0])
print(s.at['a'], s.iat[0])

# 不建议在该库下使用python原生索引方式, 防止显式与隐式冲突
```

```python
# 返回series前五行的内容
print(s.head())
# 返回series后五行的内容
print(s.tail())
# 可传入参数, 返回设定的行数, 默认为5
# 用于快捷查看元素内容
```

### 常见函数

```python
import numpy as np
# pandas会将np的缺失值及原生的None全记作NaN
s = pd.Series([10, 2, np.nan, None, 3, 4, 5], index = ['a', 'b', 'c', 'd', 'e', 'f', 'g'], name = 'Data')

s.head(3).to_frame()    # 不打印, 直接获取表格形式, 并渲染
```

```python
# 查看所有描述性信息
s.describe().to_frame()
# 几种统计方法的值都在这里, 写法与np一样, 注意计算时排除缺失值
```

```python
# 获取元素个数(忽略缺失值)
print(s.count())

# 两种获取索引的方法
print(s.keys())
print(s.index)

# 判断每一个元素是否为缺失值
s.isna().to_frame()

# 检查series的每个元素是否在给定集合中
s.isin([4, 5, 6]).to_frame()
```

```python
# 分位数写法与np有区别
s.quantile(0.25)
```

```python
# 众数, 按照出现频率计算
print(s.mode())

# 列表元素计数
print(s.value_counts())

# 按值与按索引排序
print(s.sort_values())
print(s.sort_index())
```

```python
# 去重
print(s.drop_duplicates())
# 这里对缺失值进行了去重

# 去重后的元素与个数
print(s.nunique())
```

### 6/25/26 series案例

1 学生成绩统计\
创建包含10名学生的Series, 成绩范围在50-100之间\
计算平均分 最高分 最低分 找出高于平均分的人数

```python
np.random.seed(42)
scores = pd.Series(np.random.randint(50, 101, 10), index = ['学生'+str(i) for i in range(1, 11)])

print(scores)
print(scores.mean())
print(scores.max())
print(scores.min())
mean = scores.mean()
print(len(scores[scores > mean]))    # 布尔索引筛选, 或使用.count()代替len计数
```

2 温度数据分析\
给定某城市一周内每天最高温度series\
找出温度超过30的天数\
计算平均温度 将温度从高到低排序\
找出温度变化最大的两天

```python
tem = pd.Series([28, 31, 29, 32, 30, 27, 33], index = ['周一', '周二', '周三', '周四', '周五', '周六', '周日'])

print(tem[tem > 30].count())
print(tem.mean())
print(tem.sort_values(ascending = False))    # 是否升序排列

t = tem.diff().abs()    # 获取后一个元素与前一个元素的差值
date = t.sort_values(ascending = False).keys()
print(date)
print(*date[:2])   # 这和前边显式隐式索引不一样, 这里已经选取了key()列
```

3 股票价格分析\
给定10天收盘价\
计算每日收益率, 当日 / 前日 - 1\
找出收益率最高和最低的日期\
计算波动率(标准差)

```python
prices = pd.Series([102.3, 103.5, 105.1, 104.8, 106.2, 107.0, 106.5, 108.1, 109.3, 110.2], \
                   index = pd.date_range('2023-01-01', periods = 10))    # 日期序列的函数
print(prices)
pct = prices.pct_change()    # 收益率的内置函数
print(pct)
print(pct.max(), pct.min())
print(pct.idxmax(), pct.idxmin())
print(pct.std())
```

4 销售数据分析\
计算季度平均销量, 找出销量最高月份\
计算月环比增长率\
找出连续增长超过两个月的月份\
同比增长指本月与去年本月相比, 环比指本月与上月相比

```python
sales = pd.Series([120, 135, 145, 160, 155, 170, 180, 175, 190, 200, 210, 220], \
                   index = pd.date_range('2022-01-01', periods = 12, freq = "MS"))  # 按月初生成
print(sales)
season = sales.resample("QS")    # 按季度重新采样
print(season.mean())
print(sales.idxmax())

pct = sales.pct_change(); print(pct)
pct_bool = pct > 0
pct_bool.rolling(3).sum()                                          # 宽度为3的滑动窗口求和
print(len(pct_bool[pct_bool.rolling(3).sum() > 2]))                # 布尔索引并计数
print(*pct_bool[pct_bool.rolling(3).sum() > 2].keys().tolist())    # 打印这些满足条件的标签
```

5 每小时销售数据分析\
按天重采样计算每日总销售额\
计算每天营业时间(8-22点)和非营业时间的销售额比例\
找出销售额最高的三个小时

```python
np.random.seed(42)
hourly_sales = pd.Series(np.random.randint(0, 100, 24), \
                         index = pd.date_range('2025-01-01', periods = 24, freq = 'h'))

total = hourly_sales.resample('D').sum()      # 按天重采样的总销售额
hourly_sales.between_time('8:00', '22:00')    # 筛选一段时间内Series
business = hourly_sales[(hourly_sales.index.hour <= 22) & (hourly_sales.index.hour >= 8)]    # 或采用布尔索引方式
# hourly_sales.drop(business) # 可以使用drop去掉营业时间的数据
print(business.sum() / (total - business.sum()))

print(hourly_sales.nlargest(3))    # 取出最大的几个数据, 或sort_values(ascending=False)
```

### 6/27/26 创建Dataframe
区别于Series, Dataframe为二维数据表, 可以看作多个Series拼接在一起

```python
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([6, 7, 8, 9, 10])
df = pd.DataFrame({"c1": s1, "c2": s2})    # 需使用字典创建dataframe, 或者展开如下
df1 = pd.DataFrame(
    {
        "id": [1, 2, 3, 4, 5],\
        "name": ['a', 'b', 'c', 'd', 'e'],\
        "age": [12, 14, 13, 11, 15],\
        "score": [60.5, 72, 65, 76, 76.5]
    }, index = ['r1', 'r2', 'r3', 'r4', 'r5'], columns = ["score", "id", "name", "age"]
    # 指定行索引与列顺序
)
df1
```

### 属性
与series相比增加了columns, 上文应用过了

```python
# 行索引
print(df1.index)
# 列标签
print(df1.columns)
# 值
print(df1.values)
# 维度
print(df1.ndim)
# 数据类型
print(df1.dtypes)
# 形状
print(df1.shape)
# 元素个数
print(df1.size)
# 转置
print(df1.T)
# 此时它的行索引与列标签相对调
print(df1.T.index)
```

```python
print(df1)
# 获取某行数据
print(df1.loc['r1'])
print(df1.iloc[0])
# 获取某位置数据
print(df1.at['r2', 'score'])
print(df1.iat[1, 0])
```

```python
# 获取多列数据
print(df1['score'])    # 直接写入列名或行名, 套两层会返回一个Dataframe
print(df1.score)       # 这两者的效果是相同的
print(df1[['score', 'id']])    # 获取了多列数据

# 获取多行数据
print(df1.head(2))
print(df1.tail(1))
```

```python
# 切片
df1[df1.score > 70]
df1[ (df1['score'] > 70) & (df1['id'] > 3)]

# 随机抽样
df1.sample(1)
```

### 6/29/26 Dataframe案例

案例1 学生成绩分析\
计算每个学生成绩总分和平均分\
找出数学高于90或英语高于85的\
按总分从高到低排序

```python
data = {
    '姓名': ['a', 'b', 'c', 'd', 'e'], \
    '数学': [85, 92, 78, 88, 95], \
    '英语': [90, 88, 85, 92, 80], \
    '物理': [75, 80, 88, 85, 90]
}
scores = pd.DataFrame(data)
print(scores)

scores['总分'] = scores[['数学', '英语', '物理']].sum(axis = 1)
print(scores)
scores['平均分'] = scores['总分'] / 3
print(scores)
scores['平均分'] = scores[['数学', '英语', '物理']].mean(axis = 1)    # 得到相同效果
print(scores)
```

```python
scores[ (scores['数学'] > 90) | (scores['英语'] > 85) ]
```

```python
scores.sort_values('总分', ascending = False).head(3)
```

案例2 销售数据分析\
计算销售额\
找出销售额最高的产品\
按销售额从高到低排序并输出产品信息


```python
data = {
    '产品名称': ['a', 'b', 'c', 'd'], \
    '单价': [100, 150, 200, 120], \
    '销量': [50, 30, 20, 40]
}
df = pd.DataFrame(data)
df['总销售额'] = df['单价'] * df['销量']
df.nlargest(1, columns = ['总销售额'])
df.sort_values('总销售额', ascending = False)
```

案例3 用户行为分析\
计算每位用户总消费金额\
找出消费金额最高的用户, 输出其全部信息\
计算所有用户平均消费金额,两位小数\
统计电子产品总购买量

```python
data = {
    'id': [101, 102, 103, 104, 105], \
    '用户名': ['apple', 'butter', 'charlie', 'duff', 'edward'], \
    '商品类别': ['电子产品', '服饰', '电子产品', '家居', '服饰'], \
    '商品单价': [1200, 300, 800, 150, 200], \
    '购买数量': [1, 3, 2, 5, 4]
}
df = pd.DataFrame(data)
df['总消费金额'] = df['商品单价'] * df['购买数量']
df.nlargest(1, columns = '总消费金额')
df['总消费金额'].mean()
df[df['商品类别'] == '电子产品']['购买数量'].sum()
```
