## 6/16/26 ndarray的特性
### 多维性


```python
import numpy as np
```

生成并打印零维、一维、二维ndarray数组

```python
arr = np.array(5)
print(arr)
print("arr's number of dimensions:", arr.ndim)
```

```python
arr = np.array([1, 2, 3])
print(arr)
print("arr's number of dimensions:", arr.ndim)
```

```python
arr = np.array([ [1, 2, 3], [4, 5, 6] ])
print(arr)
print("arr's number of dimensions:", arr.ndim)
```

### 同质性

放入的数据类型若不同则会被强制转化为相同数据类型

```python
arr = np.array([1, "hello"])
print(arr)
```

```python
arr = np.array([1, 2.5])
print(arr)
```
