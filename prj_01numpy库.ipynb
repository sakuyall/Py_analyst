## 6/17/26 ndarry的创建

```python
# 基础的创建方法就像前边那样
li = [[1, 2, 3], [4, 5, 6]]
arr = np.array(li, dtype = np.float64)    # 另加参数转换数据类型
print(arr, arr.dtype)
```

```python
# copy
arr1 = np.copy(arr)    # 拷贝后对arr1的修改不会影响arr原数组, 这说明内层嵌套不共享内存, 为深拷贝
arr1[0][0] = 100
print(arr, arr1, sep = "\n")
```

```python
# 预定义形状
# 全0 全1 未初始化 固定值
arr = np.zeros((2, 3)); print(arr, "这是全零数组 ")
arr = np.zeros((200, ), dtype = int); print(arr, "一行200个整型0 ")
arr = np.ones((5, 8), dtype = int); print(arr, "全一数组")
arr = np.empty((2, 3)); print(arr, "未初始化空数组执行更快")
arr = np.full((3, 4), 2026); print(arr, "填充其他内容设定")
```

```python
arr1 = np.zeros_like(arr); print(arr1)    # 输出目标数组形状的数组
arr1 = np.full_like(arr, 666); print(arr1)
```

```python
# 等差数列
arr = np.arange(1, 10, 1); print(arr)    # 与range()相同

# 等间隔数列, space形式都是包含末尾的, 下同
arr = np.linspace(0, 10, 5); print(arr)    # 设定首尾以及需要几个数
arr = np.arange(0, 11, 2.5); print(arr)    # arange这样写就可以达到同一效果, 注意不要写成"安排"

# 对数间隔数列
arr = np.logspace(0, 5, 6, base = 2); print(arr)    # 从0到5获取六个数记作pow, 依次输出base ** pow
```

### 索引与切片

```python
# 一维数组的索引与切片, 语法与寻常切片一致
arr = np.random.randint(1, 100, 30)
print(arr)
print(arr[0: 6])
print(arr[(arr > 90) & (arr < 100)])    # 通过布尔条件筛选满足条件的元素, 可用逻辑运算符
```

```python
# 二维数组的索引与切片
arr = np.random.randint(1, 100, (4, 4))
print(arr)
print(arr[1][3])    # 传统的分步索引
print(arr[1, 3])    # 元组索引, 效率更高, 仅可在numpy中使用, 在修改元素值时也采用此方法
print(arr[1: 3, 1: 3]) # 多行切片
```

1.温度数据分析\
某城市一周的最高气温为[28, 30, 29, 31, 32, 30, 29]\
计算平均气温最高气温与最低气温\
找出气温超过30的天数


```python
import numpy as np
temps = np.array([28, 30, 29, 31, 32, 30, 29])

print(f"平均气温为: {np.mean(temps):.3f}")
print(f"最高气温为: {np.max(temps):d}")
print(f"最低气温为: {np.min(temps):d}")
print(f"超过30度的天数为: {len(temps[temps > 30])}")    # 布尔类型切片

li = np.where(temps > 30, 1, 0)
print(f"超过30度的天数为: {np.cumsum(li)[-1]}")    # 使用where判断加前缀和
print(f"超过30度的天数为: {np.count_nonzero(li)}") # 或者现成的命令
```

2.学生成绩统计\
成绩为[85, 90, 78, 92, 88]\
计算成绩的平均分、中位数、标准差\
将成绩转化为百分制, 满分为10

```python
import numpy as np
score = np.array([85, 90, 78, 92, 88])
print(f"平均分为: {np.mean(score)}")
print(f"中位数为: {np.median(score)}")
print(f"标准差为: {np.std(score):.3f}")
print(score / 100 * 10)
```

4.随机数生成\
生成一个三行四列随机整数数列, 范围0-9\
计算每列中最大值和每行最小值\
将数组中所有奇数替换为-1

```python
np.random.seed(0)
arr = np.random.randint(0, 10, (3, 4))
print(arr)
print(f"每列的最大值:{np.max(arr, axis = 0)}")    # 0为列, 1为行
print(f"每列的最大值:{np.min(arr, axis = 1)}")
print(np.where(arr % 2 == 1, -1, arr))    # where替换

arr[arr % 2 == 1] = -1; print(arr)    # 布尔索引替换也可行
```

9.综合应用\
五天的销售额与成本如列表所写\
计算每天的利润\
计算利润的平均值与标准差\
找出利润最高的天数

```python
total = np.array([20, 25, 22, 30, 28])
worth = np.array([15, 18, 16, 22, 20])

get = total - worth
print(get)
print(np.mean(get))
print(f"{np.std(get):.3f}")

print(len(get[get == np.max(get)]))    # 利用布尔索引输出最大值出现的次数
```
